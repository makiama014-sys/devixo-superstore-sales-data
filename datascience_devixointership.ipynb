```python
import numpy as np
import pandas as pd

# Load dataset
df = pd.read_csv("train.csv")

print("Shape (rows, columns):", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nInfo & Data Types:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

# 1. Identify numerical and categorical columns
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()

print(f"Numerical Columns ({len(num_cols)}):", num_cols)
print(f"Categorical Columns ({len(cat_cols)}):", cat_cols)

# 2. Generate basic statistical information for numerical columns
print("\n--- Numerical Statistical Summary ---")
display(df[num_cols].describe().T)

# 3. Generate basic statistical information for categorical columns
print("\n--- Categorical Statistical Summary ---")
display(df[cat_cols].describe().T)
```

Part 2: Data cleaning

```python
df_clean = df.copy()
# 1. Check missing values before cleaning
print("1) Missing values BEFORE cleaning:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
# 2. Handle missing values appropriately
# Postal Code has 11 missing values -> fill with mode (most frequent code) & format as 5-digit string
postal_mode = df_clean["Postal Code"].mode()[0]
df_clean["Postal Code"] = (
    df_clean["Postal Code"]
    .fillna(postal_mode)
    .astype(int)
    .astype(str)
    .str.zfill(5)
)
# 3. Check and remove duplicate records
dups_count = df_clean.duplicated().sum()
print(f"\n2) Duplicate rows found: {dups_count}")
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

# 4. Identify incorrect data types & fix them
# Order Date & Ship Date should be datetime; Postal Code should be string (identifier, not numeric)
df_clean["Order Date"] = pd.to_datetime(
    df_clean["Order Date"], format="mixed", dayfirst=True, errors="coerce"
)
df_clean["Ship Date"] = pd.to_datetime(
    df_clean["Ship Date"], format="mixed", dayfirst=True, errors="coerce"
)
# Clean extra spaces in text/categorical columns
for col in df_clean.select_dtypes(include="object").columns:
  df_clean[col] = df_clean[col].astype(str).str.strip()

# Create helpful derived features for analysis
df_clean["Shipping_Days"] = (
    df_clean["Ship Date"] - df_clean["Order Date"]
).dt.days
df_clean["Order_Year"] = df_clean["Order Date"].dt.year
df_clean["Order_Month"] = df_clean["Order Date"].dt.month

# 5. Check invalid or inconsistent values
invalid_sales = (df_clean["Sales"] <= 0).sum()
invalid_dates = (df_clean["Shipping_Days"] < 0).sum()
print(f"\n3) Invalid Sales (<= 0): {invalid_sales}")
print(f"   Invalid Dates (Ship Date < Order Date): {invalid_dates}")

# 6. Verify dataset after cleaning
print("\n4) Missing values AFTER cleaning (total):", df_clean.isnull().sum().sum())
print("   Shape after cleaning:", df_clean.shape)
print("\n   Updated data types:")
print(df_clean.dtypes)
```

**Part 3: Feature Analysis **

```python
# ==========================================
# PART 3 - FEATURE ANALYSIS
# ==========================================
num_cols_part3 = ["Sales", "Shipping_Days", "Order_Year", "Order_Month"]
cat_cols_part3 = [
    "Category",
    "Sub-Category",
    "Segment",
    "Region",
    "Ship Mode",
    "State",
]

print("1) NUMERICAL FEATURES (Min, Max, Average/Mean, Median, Std & Outliers):")
for col in num_cols_part3:
  s = df_clean[col]
  q1 = s.quantile(0.25)
  q3 = s.quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5 * iqr
  upper_bound = q3 + 1.5 * iqr
  outliers = s[(s < lower_bound) | (s > upper_bound)]
  print(f"\n--- {col} ---")
  print(
      f"Min: {s.min():.2f} | Max: {s.max():.2f} | Mean (Avg): {s.mean():.2f} |"
      f" Median: {s.median():.2f} | Std: {s.std():.2f}"
  )
  print(
      f"Skewness: {s.skew():.2f} | IQR Outliers: {len(outliers)}"
      f" ({len(outliers)/len(s)*100:.2f}%)"
  )

print("\n2) CATEGORICAL FEATURES & IMPORTANT CATEGORIES:")
for col in cat_cols_part3:
  counts = df_clean[col].value_counts()
  percents = df_clean[col].value_counts(normalize=True).mul(100).round(2)
  cat_summary = pd.DataFrame({"Order_Count": counts, "Share_%": percents})
  print(f"\nTop values in {col} (Unique: {df_clean[col].nunique()}):")
  display(cat_summary.head(6))

print("\n3) SALES PERFORMANCE BY IMPORTANT CATEGORIES:")
for col in ["Category", "Segment", "Region", "Ship Mode"]:
  sales_by_cat = (
      df_clean.groupby(col)["Sales"]
      .agg(["count", "sum", "mean", "median"])
      .sort_values(by="sum", ascending=False)
  )
  print(f"\nSales summary by {col}:")
  display(sales_by_cat.round(2))
```

```python
# ==========================================
# PART 3 (FINAL PART): MOST IMPORTANT FEATURES
# ==========================================
important_cats = ["Category", "Sub-Category", "Segment", "Region", "Ship Mode"]

for col in important_cats:
  summary = (
      df_clean.groupby(col)["Sales"]
      .agg(
          Orders="count",
          Total_Sales="sum",
          Avg_Sales="mean",
          Median_Sales="median",
      )
      .sort_values(by="Total_Sales", ascending=False)
  )

  summary["Sales_Share_%"] = (
      summary["Total_Sales"] / summary["Total_Sales"].sum() * 100
  ).round(2)
  print(f"\n=== Impact of '{col}' on Sales ===")
  display(summary.round(2).head(8))

# Time-based & shipping impact
print("\n=== Impact of Order_Year on Sales ===")
display(
    df_clean.groupby("Order_Year")["Sales"]
    .agg(Orders="count", Total_Sales="sum", Avg_Sales="mean")
    .round(2)
)

print("\n=== Impact of Shipping_Days on Sales ===")
display(
    df_clean.groupby("Shipping_Days")["Sales"]
    .agg(Orders="count", Total_Sales="sum", Avg_Sales="mean")
    .round(2)
)
```

**PART 4 CORRELATION ANALYSIS**

```python
# ==========================================
# PART 4 - CORRELATION ANALYSIS
# ==========================================
import matplotlib.pyplot as plt
import seaborn as sns

# Select suitable numerical features (excluding IDs / Postal Code)
corr_cols = ["Sales", "Shipping_Days", "Order_Year", "Order_Month"]
corr_matrix = df_clean[corr_cols].corr()

print("Correlation Matrix:")
display(corr_matrix.round(4))

# Heatmap visualization
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidths=0.5,
)
plt.title("Correlation Matrix of Suitable Numerical Features")
plt.tight_layout()
plt.savefig("01_correlation_heatmap.png", dpi=300)
plt.show()
```

```python
# SETUP CELL (Run once)
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams["axes.titlesize"] = 13
plt.rcParams["axes.titleweight"] = "bold"
plt.rcParams["axes.labelsize"] = 11

df = pd.read_csv("train.csv")
df_clean = df.copy()

postal_mode = df_clean["Postal Code"].mode()[0]
df_clean["Postal Code"] = (
    df_clean["Postal Code"]
    .fillna(postal_mode)
    .astype(int)
    .astype(str)
    .str.zfill(5)
)

df_clean["Order Date"] = pd.to_datetime(
    df_clean["Order Date"], format="mixed", dayfirst=True, errors="coerce"
)
df_clean["Ship Date"] = pd.to_datetime(
    df_clean["Ship Date"], format="mixed", dayfirst=True, errors="coerce"
)

for col in df_clean.select_dtypes(include="object").columns:
  df_clean[col] = df_clean[col].astype(str).str.strip()

df_clean["Shipping_Days"] = (
    df_clean["Ship Date"] - df_clean["Order Date"]
).dt.days
df_clean["Order_Year"] = df_clean["Order Date"].dt.year
df_clean["Order_Month"] = df_clean["Order Date"].dt.month
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

dollar_fmt = mtick.StrMethodFormatter("${x:,.0f}")
print("Setup ready! Cleaned shape:", df_clean.shape)
```

**Histogram (Sales Distribution)**

```python
plt.figure(figsize=(10, 5))
ax = sns.histplot(
    df_clean[df_clean["Sales"] <= 2000]["Sales"], bins=35, kde=False
)
ax.xaxis.set_major_formatter(dollar_fmt)
plt.title("1. Histogram: Distribution of Order Sales (Orders up to $2,000)")
plt.xlabel("Order Sales ($)")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.savefig("viz_01_histogram_sales.png", dpi=300)
plt.show()
print(
    "Finding 1: Most orders cluster at lower-to-moderate dollar amounts,"
    " showing a right-skewed sales distribution."
)
```

**Bar Chart (Total Sales by Category)**

```python
cat_sales = (
    df_clean.groupby("Category")["Sales"].sum().sort_values(ascending=False)
)
plt.figure(figsize=(10, 5))
ax = cat_sales.plot(kind="bar")
ax.yaxis.set_major_formatter(dollar_fmt)
plt.title("2. Bar Chart: Total Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("viz_02_bar_category_sales.png", dpi=300)
plt.show()
print(
    "Finding 2: Technology generates the highest total sales, followed by"
    " Furniture and Office Supplies."
)
```

**Count Plot (Order Count by Ship Mode)**

**Pie Chart (Revenue Share by Segment)**
